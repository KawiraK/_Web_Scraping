## Step 4: Parse the HTML

`response.text` is one enormous string. `BeautifulSoup` turns it into a tree you can actually
search through by tag and class, instead of doing string matching yourself.


```python
soup = BeautifulSoup(response.text, "html.parser")

# Print the first 800 characters so you can see the structure you are about to search through.
# This should look like real, readable HTML, not an error page or a login wall.
print(soup.prettify()[:800])
```

## Step 5: Find every repeating item on the page

This is the main challenge in this assignment. Every item on your page (book, quote, listing,
whatever you picked) is wrapped in the same repeating tag and class, the one you found with
DevTools in Step 2. `soup.find_all()` returns every element matching that tag and class as a
list.


```python
# TODO: replace both ___ with the real tag name and class name from your own page.
# Example shape only, your real values will be different: soup.find_all("article", class_="product_pod")
items = soup.find_all("div", class_="country")

print(len(items))

# TODO: open the actual page in your browser and count the items by eye (or estimate for a long
# page). Does your count roughly match? If items is empty or the count looks wrong, your tag or
# class name is not matching what you think it is. Go back to DevTools and check again before
# moving on, extracting fields from an empty list will just waste the rest of this notebook.
```

## Step 6: Extract the fields you actually care about
